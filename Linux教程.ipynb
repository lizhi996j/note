### history [选项] [参数]

#### 与history有关的环境变量
- HISTSIZE: 加载到内存中的历史命令记录的条数
- HISTFILE: 记录历史命令的文件
- HISTFILESIZE: 记录历史命令的文件的大小
- HISTCONTROL: 控制历史命令的记录方式
    - ignorespace: 以空格开头的命令不记录。
    - ignoredups: 与上一条相同的命令不记录（仅跳过连续重复）。
    - ignoreboth: 相当于 ignorespace:ignoredups 的简写。
    - erasedups: 记录当前命令前，删除历史中所有相同的旧条目（只保留最新一次）。
    一个例子： HISTCONTROL=ignorespace:ignoredups:erasedups


````python
%%bash
source ~/.bashrc
echo ${HISTSIZE}
# echo $HISTFILE
# echo $HISTFILESIZE
# echo $HISTCONTROL
````

#### source 的主要用途
加载环境变量

1. 加载环境变量和配置文件
- 常用于加载配置文件（如 .bashrc、.bash_profile）
- 示例: `source ~/.bashrc` 
  重新加载当前用户的 .bashrc 文件，应用其中定义的环境变量、别名等配置

2. 执行脚本并保留变量
- 使用 source 执行脚本时，变量会保留在当前 Shell 会话中
- 示例:
  ```bash
  # example.sh
  MY_VAR="Hello, world!"
  
  # 使用 source
  source example.sh
  echo $MY_VAR  # 输出: Hello, world!
  ```
- 直接运行脚本（如 bash example.sh）则变量不会保留

3. 运行设置脚本
- 用于配置开发环境或虚拟环境
- 示例: `source venv/bin/activate` (Python虚拟环境)

4. source 与直接执行脚本的区别:
| 功能 | source 执行脚本 | 直接执行脚本 |
|------|----------------|--------------|
| 运行环境 | 当前 Shell | 启动新的子 Shell |
| 变量保留 | 是 | 否 |
| 适用场景 | 修改当前Shell环境 | 独立执行脚本 |


#### /opt和/usr区别：(usr:Unix System Resources,opt:optional)
/usr 是系统标准软件的"共享仓库"，而 /opt 是"外来的大软件"专属地盘。

/usr vs /opt 对比表：
| 特性 | /usr | /opt |
|------|------|------|
| 用途 | 系统自带 & 标准包管理器安装的软件（如 APT 安装的） | 外部提供的独立大软件（通常是第三方预编译包） |
| 结构 | 拥有多个子目录，如 /usr/bin, /usr/lib, /usr/share 等 | 每个软件一个子目录，如 /opt/google/chrome/ |
| 管理方式 | 通常由包管理器（如 apt、dpkg）管理 | 通常是用户手动解压或安装的 |
| 设计目标 | 提供给所有用户共享的软件库和工具 | 容纳"非标准"、自包含、不会影响系统的独立应用 |
| 示例 | /usr/bin/vim, /usr/lib/libc.so.6 | /opt/pycharm, /opt/Matlab, /opt/Google/chrome |
#### 桌面图标的位置：
`/usr/share/applications/`
#### `/bin`和`/usr/bin`和`/usr/local/bin`的区别：
- `/bin`：存放的是系统启动和运行最基本的可执行程序，在系统尚未挂载 /usr、网络不可用、图形界面没启动之前，系统就必须能执行这些命令；，如 `ls`, `cp`, `mv`,`rm` 等。
- `/usr/bin`：/usr/bin 是 Linux 中用于存放非基本命令的用户级可执行程序（binary executables）的目录，如 `ssh`,`vim`,`code`(VScode) 等。
- `/usr/local/bin`：/usr/local 是为“本地手动安装的软件”预留的目录，用于区别于系统软件包管理器（如 apt、yum）自动安装的软件。
自定义安装的是指：解压 .tar.gz 到某路径，下载 .AppImage、.run、.sh 安装脚本， 通过./configure && make && sudo make install 安装。

## Linux 系统目录结构
- `/bin`:  
`/bin`：存放的是系统启动和运行最基本的可执行程序，在系统尚未挂载 /usr、网络不可用、图形界面没启动之前，系统就必须能执行这些命令；，如 `ls`, `cp`, `mv`,`rm` 等。
    - 与/usr的区别是：  
    	/bin是最基本的系统命令（启动、修复系统必需的），即使系统只挂载了根分区 /，也能用；
        /usr是用户级别的程序和数据（大部分命令、库文件），需要系统进入正常状态才能使用；
- `/usr`： (unix system resource),   
内容：  
`/usr/bin` : 大多数普通命令（比如 python, gcc, vim）  
`/usr/lib` : 库文件, 类似于python包；  以.so结尾，windows中以.dll结尾  
`/usr/share` : 共享数据（字体、图标、文档）  
`/usr/local/bin` : 用于存放“手动/本地安装”的软件库文件（不由系统包管理器管理,也就是不由apt、yum等安装），比如  
    - 从源码编译并安装：./configure --prefix=/usr/local && make && sudo make install
    - 厂商脚本或手动把库拷进 /usr/local/lib
    - 自行解压二进制到 /usr/local 或 /opt
- `dev`:管理设备
- `var`: 存放可写、随运行变化的可变数据（会增长/轮转/清理）。比如日志、缓存、队列、PID/lock、服务状态与数据库
    - `var`与`/usr/share`的区别：
        `/usr/share`只读、架构无关、静态数据（安装时放入，运行期不改）。
        `/var`可写、架构相关、动态数据（运行期增长/轮转/清理）。


## 文件的基本属性

### 查看文件属性：
- 命令：ls -l
    - 第一个字符代表这个文件是目录、文件或链接文件等等。
        - 当为 d 则是目录
        - 当为 - 则是文件；
        - 若是 l 则表示为链接文档(link file)；
        - 若是 b 则表示为装置文件里面的可供储存的接口设备(可随机存取装置)；
        - 若是 c 则表示为装置文件里面的串行端口设备，例如键盘、鼠标(一次性读取装置)。
    - 之后以三个为一组，表示属主，属组和其它用户权限
    - 其余输出详细含义见[菜鸟教程](https://www.runoob.com/linux/linux-file-attr-permission.html)

### 改变文件属组
- 命令： chgrp [-R] 属组名 文件名

### 改变文件属主
- 命令： chown   
chown [–R] 所有者 文件名  
chown [-R] 所有者:属组名 文件名

### 改变文件权限
- 命令 chmod  
- 方法一：字母法
chmod [who][operator][permission] 文件名
    - who:
        - u: 属主
        - g: 属组
        - o: 其它用户
        - a: 所有人
    - operator:
        - +: 添加权限
        - -: 移除权限
    - permission:
        - r: 读权限
        - w: 写权限
        - x: 执行权限

- 方法二：数字法
chmod [mode=] 数字 文件名
    - 数字权限说明：
        - r(读) = 4
        - w(写) = 2  
        - x(执行) = 1
    - 每种身份(owner/group/others)各自的三个权限(r/w/x)分数累加形成三位十以内数字

#### 字符串

    - 双引号里可以有转义字符(比如\n,\t)
    - 也可以有命令替换： $(command) 或 `command`
- 拼接字符串：
    - 使用双引号拼接
    - 使用单引号拼接
- 注：与python相同，单引号字符串里不能有单引号，双引号字符串里不能有双引号；  
而且python里面如果两个引号连在一起也会拼接

````python
%%bash
your_name="runoob"1
echo $your_name
# 单引号字符串
str1='hello, "$your_name"'
echo $str1
# str2='It's fine' 报错


## 单引号拼接
str2='It is fine'
echo $str1$str2 # echo会用空格将后面的参数链接起来
str3="$str1 $str2" # 注意str3=$str1 $str2是错误的，因为会把赋值之后的命令理解成另一句命令；


# str3=$str1 $str2会把空格前面理解成赋值语句，后面理解成带着环境变量str3去执行命令，
#比如： PATH=/tmp ls 用临时 PATH 执行 ls，并不会永久改 PATH。
echo "str3的值是${str3}"


# 使用双引号拼接
greeting="hello, "$your_name" !" #是拼接，"hello, "被当作第一部分
greeting_1="hello, ${your_name} !"
echo $greeting  $greeting_1
````

# 踩坑

### shebang的重要性；

- 原因：对于shell脚本，有bash和sh两种解释器；
- 遇到问题：对于history命令，sh会报错，bash会输出为0；但在命令行以可执行程序运行没有shebang的踩坑_0.sh时却输出了history的结果，用sh编译会报错，用bash编译会输出history的结果为0；为什么.sh会输出history的结果？
- 回答：当运行./shell_script/踩坑_0.sh时，由于一开始的版本没有shebang, 会使用sh编译， 但会报错，报错之后bash有回退机制，脚本在当前终端的 Bash 会话中执行，而不是启动新子进程（相当于../shell_script/踩坑_0.sh）；
